## 4. Evidencia climática de la amenaza (de su capítulo 1)

La amenaza no se inventa: se apoya en los datos. Esta celda calcula, para la zona de la Práctica 1, los indicadores que sostienen la calificación de cada amenaza. La **sugerencia** es solo un punto de partida para discutir; la calificación final es suya y se justifica con esta evidencia **y** con lo observado en el sitio.

```python
D = pd.read_csv(DATOS / f"serie_diaria_{S}.csv", index_col="fecha", parse_dates=True)
M = pd.read_csv(DATOS / f"serie_mensual_{S}.csv", index_col="fecha", parse_dates=True)
META1 = json.loads((P1 / "IA1_03_metadatos.json").read_text(encoding="utf-8"))
ANIOS_NINO = META1["anios_nino_temporada"]
FUENTES = [f for f in ("ERA5", "CORe") if f"{f}_pr" in D]
EVID, IND = {}, {}
for f in FUENTES:
    ip = X.precipitacion(D[f"{f}_pr"], (1981, 2010))
    it = X.temperatura(D[f"{f}_tmax"], D[f"{f}_tmin"], (1981, 2010)) if f"{f}_tmax" in D else pd.DataFrame()
    IND[f] = pd.concat([ip, it], axis=1)
    rx = ip["Rx1day"].dropna()
    ret = X.retorno(rx, (10, 25))
    temp = enso.temporada(M[f"{f}_pr"], (12, 1, 2, 3, 4, 5), "sum").dropna()
    nin = temp[temp.index.isin(ANIOS_NINO)]; otros = temp[~temp.index.isin(ANIOS_NINO)]
    rx_n = rx[rx.index.isin(ANIOS_NINO)]; rx_o = rx[~rx.index.isin(ANIOS_NINO)]
    e = {"Lluvia dic–may, años sin El Niño (mm)": otros.mean(), "Lluvia dic–may, años de El Niño (mm)": nin.mean(),
         "Factor El Niño (lluvia dic–may)": nin.mean() / otros.mean(),
         "Rx1day medio (mm)": rx.mean(), "Rx1day medio en El Niño (mm)": rx_n.mean(), "Rx1day medio sin El Niño (mm)": rx_o.mean(),
         "Rx1day de 10 años (mm)": ret.loc[10, "nivel"], "Rx1day de 25 años (mm)": ret.loc[25, "nivel"],
         "Rx5day medio (mm)": ip["Rx5day"].mean(), "P95 de días lluviosos (mm/día)": ip.attrs["umbrales"]["P95_mm"],
         "R95pTOT, tendencia (%/década)": X.tendencia(ip["R95pTOT"])["sen_por_decada"],
         "CDD medio (días)": ip["CDD"].mean(), "CDD, tendencia (días/década)": X.tendencia(ip["CDD"])["sen_por_decada"],
         "PRCPTOT, tendencia (mm/década)": X.tendencia(ip["PRCPTOT"])["sen_por_decada"]}
    if not it.empty:
        t = X.tendencia(it["TX90p"])
        e.update({"TX90p medio (% de días)": it["TX90p"].mean(), "TX90p, tendencia (%/década)": t["sen_por_decada"],
                  "TX90p, p corregido": t["p_valor"], "TXx medio (°C)": it["TXx"].mean()})
    EVID[f] = e
EVID = pd.DataFrame(EVID).astype(float).round(2)
display(EVID)
f0 = FUENTES[0]
SUG = {"inundacion": RG.sugerencia_lluvia(EVID.loc["Factor El Niño (lluvia dic–may)", f0]),
       "movimiento_masa": RG.sugerencia_lluvia(EVID.loc["Factor El Niño (lluvia dic–may)", f0]),
       "costera": 0, "sequia": 0,
       "calor": RG.sugerencia_calor(EVID.loc["TX90p, tendencia (%/década)", f0]) if "TX90p, tendencia (%/década)" in EVID.index else 0}
print(f"Años de El Niño usados (de IA1_03): {ANIOS_NINO}")
print("Sugerencias orientativas (0 = sin dato climático: califique con lo observado en campo):")
for k, v in SUG.items():
    print(f"  {RG.AMENAZAS[k]}: {v}")
print("\nLa movilidad en masa depende además de la pendiente y el suelo; la amenaza costera, del oleaje y del nivel del mar,")
print("que estos datos no miden: califíquelas con la evidencia del sitio, la historia de eventos y el AR6.")
```

## 5. Calificación de las amenazas (1–5; 0 = no aplica en el sitio)

| Valor | Descriptor |
|---|---|
| 1 | No hay registros ni evidencia física |
| 2 | Evidencia débil o eventos muy raros |
| 3 | Eventos ocasionales; evidencia en el sitio (marcas, relatos) |

## 7. Índice de riesgo y coherencia

```python
AMEN = {"inundacion": A_INUNDACION, "movimiento_masa": A_MOVIMIENTO_MASA, "costera": A_COSTERA, "sequia": A_SEQUIA, "calor": A_CALOR}
JUST_A = {"inundacion": J_INUNDACION, "movimiento_masa": J_MOVIMIENTO_MASA, "costera": J_COSTERA, "sequia": J_SEQUIA, "calor": J_CALOR}
EXPO = {"personas": E_PERSONAS, "viviendas": E_VIVIENDAS, "servicios": E_SERVICIOS, "economia": E_ECONOMIA, "ecosistemas": E_ECOSISTEMAS}
SENS = {"vivienda_precaria": S_VIVIENDA_PRECARIA, "poblacion_vulnerable": S_POBLACION_VULNERABLE, "dependencia_clima": S_DEPENDENCIA_CLIMA,
        "degradacion": S_DEGRADACION, "drenaje": S_DRENAJE}
CAPA = {"alerta": C_ALERTA, "acceso": C_ACCESO, "planificacion": C_PLANIFICACION, "recursos": C_RECURSOS}
COMP = RG.componentes(EXPO, SENS, CAPA)
RIESGO = RG.indice(AMEN, COMP)
RIESGO["justificacion"] = RIESGO.clave.map(JUST_A)
CRIT = RG.factor_dominante(EXPO, SENS, CAPA)
print("Componentes (1–5): " + " · ".join(f"{k.replace('_', ' ')} {v:.2f}" for k, v in COMP.items()))
display(RIESGO[["amenaza", "A", "E", "V", "riesgo", "clase"]])
print(f"Indicadores críticos: exposición «{RG.EXPOSICION[CRIT['exposicion']]}», sensibilidad «{RG.SENSIBILIDAD[CRIT['sensibilidad']]}», "
      f"capacidad más débil «{RG.CAPACIDAD[CRIT['capacidad']]}»")

avisos = []
for k, a in AMEN.items():
    if a and not JUST_A[k].strip():
        avisos.append(f"La amenaza «{RG.AMENAZAS[k]}» no tiene justificación.")
    if SUG.get(k) and a and abs(a - SUG[k]) >= 2:
        avisos.append(f"«{RG.AMENAZAS[k]}»: calificó {a} y la evidencia climática sugiere {SUG[k]}. Explique la diferencia con lo observado en el sitio.")
for nombre, j in (("exposición", J_EXPOSICION), ("sensibilidad", J_SENSIBILIDAD), ("capacidad", J_CAPACIDAD)):
    if not j.strip() or j.startswith("Cuántas viviendas"):
        avisos.append(f"Falta la justificación de la {nombre}.")
if len(set(EXPO.values()) | set(SENS.values()) | set(CAPA.values())) == 1:
    avisos.append("Todas las calificaciones son iguales: revise si reflejan el sitio.")
print("\n" + ("Sin observaciones de coherencia." if not avisos else "Revise antes de entregar:\n- " + "\n- ".join(avisos)))
```
