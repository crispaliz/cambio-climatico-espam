## 4. Pesos de cada celda

El promedio de la zona pondera cada celda por la **fracción de su área que cae dentro del límite** oficial y por el coseno de la latitud. Una celda partida por el borde cuenta en proporción a lo que tiene dentro; así el resultado no depende de si el centro de la celda cae o no en la zona, lo que importa mucho en cantones pequeños.

**CORe** tiene celdas de ~1,9°: se aplica el mismo método con sus propias celdas, sin interpolar. Si la zona cae en una o dos celdas de CORe, su serie es la de esas celdas: representa una escala mucho mayor que la de ERA5.

**Viento:** la **velocidad** (`ws`) es el promedio de las velocidades horarias (ERA5) o de 6 h (CORe diario); la **dirección** (`wd`) sale del viento medio vectorial (u, v). Con `CORE_DIARIO = False` no hay velocidades sub-mensuales de CORe y `ws` es la del vector medio, algo menor.

```python
FR_E = limites.fracciones(GEOM, E_M["lat"], E_M["lon"], 0.25)
C_REF = C_M if C_M is not None else C_D
FR_C = limites.fracciones(GEOM, C_REF["lat"], C_REF["lon"], 1.875) if C_REF is not None else None
print(f"ERA5: {int((FR_E > 0).sum())} celdas con peso · CORe: {0 if FR_C is None else int((FR_C > 0).sum())} celdas con peso")

fig, ax = plt.subplots(1, 2 if FR_C is not None else 1, figsize=(10, 4.5), squeeze=False)
FR_E.where(FR_E > 0).plot(ax=ax[0, 0], cmap="YlOrBr", vmin=0, vmax=1, cbar_kwargs={"label": "fracción dentro"})
ax[0, 0].set_title("ERA5 (0,25°)")
if FR_C is not None:
    FR_C.where(FR_C > 0).plot(ax=ax[0, 1], cmap="YlOrBr", vmin=0, vmax=1, cbar_kwargs={"label": "fracción dentro"})
    ax[0, 1].set_title("CORe (~1,9°)")
for a in ax[0]: gpd.GeoSeries([GEOM]).boundary.plot(ax=a, color="k", lw=0.8)
fig.suptitle(f"Pesos de las celdas en {Z['nombre']}"); fig.tight_layout(); plt.show()

VARS = ["tmed", "tmax", "tmin", "pr", "hr", "q", "ps", "u10", "v10", "ws"]
def zona_media(ds, fuente, frac):
    df = pd.DataFrame({v: limites.promedio_ponderado(ds[v], frac).to_series() for v in VARS if v in ds})
    if {"u10", "v10"} <= set(df):
        ws_vec, df["wd"] = V.viento(df["u10"], df["v10"])
        if "ws" not in df: df["ws"] = ws_vec
    return df.add_prefix(f"{fuente}_")
```

## 5. Serie mensual

Los mensuales se calcularon con los diarios: Tmax y Tmin mensuales son el promedio de las máximas y mínimas diarias. Solo si eligió `CORE_DIARIO = False` el mensual de CORe viene de los promedios mensuales de NOAA y no trae Tmax ni Tmin.

```python
partes = [zona_media(E_M, "ERA5", FR_E)]
if C_M is not None: partes.append(zona_media(C_M, "CORe", FR_C))
for p in partes: p.index = p.index.to_period("M").to_timestamp()
MENSUAL = pd.concat(partes, axis=1).round(3); MENSUAL.index.name = "fecha"
display(MENSUAL.describe().round(2).T)
```

## 6. Serie diaria

Cada fuente define el día de forma distinta: ERA5 de 00 a 24 h hora local (o desde la hora elegida en `CR01_descarga_reanalisis`); CORe de 01:00 a 01:00; las estaciones del INAMHI suelen medir la lluvia de 07:00 a 07:00. Se declara al comparar valores diarios.

```python
partes = []
if E_D is not None: partes.append(zona_media(E_D, "ERA5", FR_E))
if C_D is not None: partes.append(zona_media(C_D, "CORe", FR_C))
DIARIA = pd.concat(partes, axis=1).round(3) if partes else None
if DIARIA is not None:
    DIARIA.index.name = "fecha"; display(DIARIA.describe().round(2).T)
```

## 7. Primera mirada a las series mensuales

```python
paneles = [("tmed", "T media (°C)"), ("pr", "P (mm/mes)"), ("hr", "HR (%)"), ("q", "q (g/kg)"), ("ps", "Presión (hPa)"), ("ws", "Viento (m/s)")]
fig, ax = plt.subplots(len(paneles), 1, figsize=(12, 13), sharex=True)
for a, (v, et) in zip(ax, paneles):
    for f, c in (("ERA5", "#C2531F"), ("CORe", "#4C78A8")):
        if f"{f}_{v}" in MENSUAL: a.plot(MENSUAL.index, MENSUAL[f"{f}_{v}"], color=c, lw=0.9, label=f)
    a.set_ylabel(et); a.legend(loc="upper right", fontsize=8)
fig.suptitle(f"{Z['nombre']}: series mensuales {ANIO_INI}–{ANIO_FIN}"); fig.tight_layout()
fig.savefig(RES / f"CR02_series_{S}.png", dpi=200); plt.show()
```
