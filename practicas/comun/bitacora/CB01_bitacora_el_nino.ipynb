## 3. Índices de ENSO

- **ICEN** (Índice Costero El Niño): media móvil de 3 meses de la anomalía de temperatura del mar en Niño 1+2, frente a la costa. Aquí se calcula con los datos de NOAA (ERSSTv5, base 1991–2020); el **valor oficial** lo publica el IGP del Perú y puede diferir en unas centésimas: contrástelo.
- **ONI**: lo mismo en Niño 3.4, en el Pacífico central (NOAA CPC).

Los índices van atrasados: el último valor corresponde a uno o dos meses antes de la consulta.

```python
ONI = enso.oni(DATOS); ICEN = enso.icen(DATOS)
for k, v in {"ICEN": ICEN, "ONI": ONI}.items():
    u = v.dropna().tail(4)
    print(f"{k}: " + " · ".join(f"{i:%b %Y} {x:+.2f} °C" for i, x in u.items()))
print(f"Categoría del último ICEN: {enso.categoria_icen(float(ICEN.dropna().iloc[-1]))}")

fig, ax = plt.subplots(figsize=(11, 3.6))
for k, v, c in (("ICEN (Niño 1+2)", ICEN, "#C2531F"), ("ONI (Niño 3.4)", ONI, "#4C78A8")):
    v = v.loc["2023-01":]
    ax.plot(v.index, v.values, "o-", ms=3, color=c, label=k)
ax.axhline(enso.UMBRALES_ICEN["cálida débil"], color="#C2531F", ls="--", lw=.7, label="umbral cálido ICEN (+0,4)")
ax.axhline(enso.UMBRAL_ONI, color="#4C78A8", ls=":", lw=.7, label="umbral cálido ONI (+0,5)")
ax.axhline(0, color="k", lw=.5); ax.set_ylabel("°C"); ax.legend(fontsize=8, ncol=2)
ax.set_title("Índices de ENSO desde 2023 (preliminares los últimos meses)", fontsize=10)
CARPETA_BIT.mkdir(parents=True, exist_ok=True)
fig.tight_layout(); fig.savefig(CARPETA_BIT / f"indices_enso_{S}.png", dpi=150); plt.show()
```

## 4. Incidentes del ECU 911 en su zona desde la consulta anterior

Incidentes climáticos de gestión de riesgos (inundación, deslizamiento, vendaval, incendio forestal, «amenaza natural» sin detalle…) de su provincia o cantón. Son **días con atención coordinada, no desastres verificados**, y el portal los publica con uno o dos meses de retraso: si no aparece nada reciente, complete la columna con el SitRep de la SNGR, el GAD o la prensa local.

```python
INCIDENTES = ""
if USAR_ECU911:
    desde = pd.Timestamp(ANTERIOR or HOY - datetime.timedelta(days=30))
    try:
        inc = ecu911.incidentes(carpeta=BASE / "CC_datos" / "_ecu911", desde=f"{desde:%Y-%m}", verbose=False)
        zona = ecu911.de_zona(inc, Z["codigo"])
        ultimo = f"{inc['fecha'].max():%d/%m/%Y}" if len(inc) else "sin datos"
        INCIDENTES = bitacora.resumen_ecu911(zona, desde)
        print(f"Último día publicado por el ECU 911 en Datos Abiertos: {ultimo}")
        if INCIDENTES:
            print(INCIDENTES)
            display(zona[zona["fecha"] > desde].groupby(["tipo", "canton"]).size().rename("incidentes").to_frame())
        else:
            print("Sin incidentes climáticos publicados para su zona desde la consulta anterior. Complete con SitRep, GAD o prensa.")
    except Exception as err:
        print(f"No se pudo consultar el ECU 911 ahora ({err}). Complete la columna a mano o vuelva a ejecutar más tarde.")
else:
    print("ECU 911 desmarcado: complete «incidentes_zona» con SitRep, GAD o prensa.")
```
