## 4. Ciclo anual 1995–2014: lo normal y lo extremo

Para gestión de riesgos no basta el promedio: importa **qué tan lejos puede ir un mes**. Además de la media se muestran el rango entre los percentiles 10 y 90 (8 de cada 10 años caen dentro) y el valor máximo registrado en cada mes.

```python
R = M.loc[f"{REF_INI}":f"{REF_FIN}"]
CICLO = {}
fig, ax = plt.subplots(1, 2, figsize=(14, 4.6))
x = np.arange(12)
for k, f in enumerate(FUENTES):
    g = R[f"{f}_pr"].groupby(R.index.month)
    c = pd.DataFrame({"media": g.mean(), "p10": g.quantile(.1), "p90": g.quantile(.9), "max": g.max(),
                      "cv_%": 100 * g.std() / g.mean()})
    CICLO[f] = c
    d = (k - 0.5) * 0.38
    ax[0].bar(x + d, c["media"], 0.38, color=COL[f], alpha=.75, label=f"{f}: media")
    ax[0].errorbar(x + d, c["media"], yerr=[c["media"] - c["p10"], c["p90"] - c["media"]], fmt="none", ecolor="k", capsize=2, lw=.8)
    ax[0].plot(x + d, c["max"], "v", color=COL[f], ms=5, label=f"{f}: máximo")
    for v, est in (("tmax", "-"), ("tmed", "--"), ("tmin", ":")):
        if f"{f}_{v}" in R:
            ax[1].plot(x, R[f"{f}_{v}"].groupby(R.index.month).mean(), est, color=COL[f], label=f"{f} {v}")
ax[0].set_xticks(x, MESES); ax[0].set_ylabel("mm/mes"); ax[0].set_title("Precipitación: media, P10–P90 y máximo"); ax[0].legend(fontsize=8)
ax[1].set_xticks(x, MESES); ax[1].set_ylabel("°C"); ax[1].set_title("Temperatura máxima, media y mínima"); ax[1].legend(fontsize=7, ncol=2)
fig.suptitle(f"{Z['nombre']}: ciclo anual {REF_INI}–{REF_FIN}"); fig.tight_layout()
fig.savefig(RES / f"RD1_01_ciclo_anual_{S}.png", dpi=300); plt.show()
for f in FUENTES:
    c = CICLO[f]; c.index = MESES
    humedos = [MESES[i] for i in range(12) if c["media"].iloc[i] > c["media"].sum() / 12]
    print(f"{f}: P anual {c['media'].sum():.0f} mm · meses sobre el promedio mensual: {', '.join(humedos)} · "
          f"mes más variable: {c['cv_%'].idxmax()} (CV {c['cv_%'].max():.0f} %)")
    display(c.round(1).T)
```

## 5. Los índices de ENSO

Se descargan de NOAA y se guarda una copia junto a sus datos. El **ICEN** mira el Pacífico frente a la costa (Niño 1+2), donde nace El Niño costero que más afecta al litoral, como en 2017; el **ONI**, el Pacífico central (Niño 3.4), que domina la teleconexión de gran escala. Pueden no coincidir: 2017 fue El Niño costero con ONI neutro, y algunos eventos centrales casi no calientan la costa.

La celda muestra además el estado más reciente: el evento en curso entra en la bitácora de El Niño, separado del análisis climatológico porque sus datos son preliminares.

```python
ONI = enso.oni(DATOS); ICEN = enso.icen(DATOS)
IDX = {"ONI": ONI, "ICEN": ICEN}
EVENTOS = {k: enso.eventos(v, k) for k, v in IDX.items()}
FASE = {k: enso.fases(v, k) for k, v in IDX.items()}

fig, ax = plt.subplots(2, 1, figsize=(13, 6), sharex=True)
for a, (k, v) in zip(ax, IDX.items()):
    v = v.loc[f"{ANIO_INI}":]
    a.fill_between(v.index, 0, v, where=v > 0, color="#C2531F", alpha=.6, step="mid")
    a.fill_between(v.index, 0, v, where=v < 0, color="#4C78A8", alpha=.6, step="mid")
    u = enso.UMBRAL_ONI if k == "ONI" else enso.UMBRALES_ICEN["cálida débil"]
    a.axhline(u, color="k", lw=.6, ls="--"); a.axhline(-enso.UMBRAL_ONI if k == "ONI" else enso.UMBRALES_ICEN["fría débil"], color="k", lw=.6, ls="--")
    a.set_ylabel(f"{k} (°C)"); a.set_title(f"{k}: últimos valores {', '.join(f'{i:%b %Y} {x:+.2f}' for i, x in v.tail(3).items())}", fontsize=9)
fig.tight_layout(); fig.savefig(RES / f"RD1_01_indices_enso.png", dpi=200); plt.show()
print(f"Episodios según {INDICE_ENSO} desde {ANIO_INI}:")
display(EVENTOS[INDICE_ENSO][EVENTOS[INDICE_ENSO]["inicio"] >= str(ANIO_INI)])
```

## 6. Fase de cada mes y anomalías

Cada mes se clasifica como **El Niño**, **La Niña** o **Neutro** según `INDICE_ENSO` y su regla de persistencia (ICEN: 3 meses seguidos; ONI: 5 trimestres seguidos). La anomalía de cada mes se calcula contra su promedio en la base 1991–2020 (la misma de los índices de ENSO): en °C para la temperatura y en **porcentaje** para la lluvia, porque 50 mm de más no significan lo mismo en marzo que en agosto.

```python
F = FASE[INDICE_ENSO].reindex(M.index)
BASE_M = M.loc[f"{ANOM_INI}":f"{ANOM_FIN}"]
CLIM = BASE_M.groupby(BASE_M.index.month).mean()
ANOM = M - CLIM.reindex(M.index.month).set_axis(M.index)
for f in FUENTES:
    ANOM[f"{f}_pr_%"] = 100 * ANOM[f"{f}_pr"] / CLIM[f"{f}_pr"].reindex(M.index.month).values
ANOM["fase"] = F
print(ANOM["fase"].value_counts().to_string())
```

## 7. Compuestos mensuales por fase

El **compuesto** es el promedio de las anomalías de todos los meses de una misma fase, mes por mes del año. Muestra **cuándo** del año actúa ENSO en su zona. Los puntos marcan los meses en que la diferencia con los años neutros es significativa (prueba de Mann-Whitney, 5 %); con pocos eventos, un compuesto no significativo no prueba que no haya efecto.

```python
from scipy.stats import mannwhitneyu
FASES = ["El Niño", "Neutro", "La Niña"]; CF = {"El Niño": "#C2531F", "Neutro": "0.5", "La Niña": "#4C78A8"}
COMP = {}
fig, ax = plt.subplots(len(FUENTES), 2, figsize=(14, 3.8 * len(FUENTES)), squeeze=False)
for i, f in enumerate(FUENTES):
    for j, (v, et) in enumerate(((f"{f}_pr_%", "Anomalía de precipitación (%)"), (f"{f}_tmed", "Anomalía de T media (°C)"))):
        tabla = ANOM.pivot_table(index=ANOM.index.month, columns="fase", values=v, aggfunc="mean").reindex(columns=FASES)
        n = ANOM.pivot_table(index=ANOM.index.month, columns="fase", values=v, aggfunc="count").reindex(columns=FASES)
        COMP[v] = tabla
        a = ax[i, j]
        for fa in FASES:
            a.plot(x, tabla[fa], "o-", color=CF[fa], label=f"{fa} (n≈{int(n[fa].mean()) if n[fa].notna().any() else 0} por mes)")
        for m in range(1, 13):
            sel = ANOM[ANOM.index.month == m]
            a_, b_ = sel.loc[sel.fase == "El Niño", v].dropna(), sel.loc[sel.fase == "Neutro", v].dropna()
            if len(a_) >= 3 and len(b_) >= 3 and mannwhitneyu(a_, b_).pvalue < 0.05:
                a.plot(m - 1, tabla.loc[m, "El Niño"], "o", ms=12, mfc="none", mec="k")
        a.axhline(0, color="k", lw=.5); a.set_xticks(x, MESES, fontsize=8); a.set_title(f"{f}: {et}", fontsize=10); a.legend(fontsize=7)
fig.suptitle(f"{Z['nombre']}: compuestos por fase de ENSO ({INDICE_ENSO}), {ANIO_INI}–{ANIO_FIN}"); fig.tight_layout()
fig.savefig(RES / f"RD1_01_compuestos_{S}.png", dpi=300); plt.show()
display(pd.concat({k: t.round(1) for k, t in COMP.items()}, axis=1).set_axis(MESES))
```

## 8. La temporada lluviosa y la intensidad del evento

Lo que define el riesgo es el **acumulado de la temporada** y su relación con la **intensidad** del evento. Cada punto es un año: en el eje x el índice medio de la temporada y en el y la lluvia de la temporada en porcentaje de su promedio. Se rotulan los años más lluviosos.

```python
from scipy.stats import spearmanr
TEMP = pd.DataFrame({INDICE_ENSO: enso.temporada(IDX[INDICE_ENSO], MESES_T, "mean")})
for f in FUENTES:
    s = enso.temporada(M[f"{f}_pr"], MESES_T)
    base = s.loc[ANOM_INI:ANOM_FIN].mean()
    TEMP[f"{f}_mm"], TEMP[f"{f}_%"] = s, 100 * s / base
TEMP = TEMP.dropna(subset=[f"{FUENTES[0]}_mm"])
def cat(v):
    if INDICE_ENSO == "ICEN": return enso.categoria_icen(v)
    return next((("cálido " if v > 0 else "frío ") + n for u, n in enso.INTENSIDAD_ONI if abs(v) >= u), "neutro")
TEMP["categoría"] = TEMP[INDICE_ENSO].map(cat)
fig, ax = plt.subplots(1, 2, figsize=(14, 4.8))
for f in FUENTES:
    r, p = spearmanr(TEMP[INDICE_ENSO], TEMP[f"{f}_%"], nan_policy="omit")
    ax[0].scatter(TEMP[INDICE_ENSO], TEMP[f"{f}_%"], color=COL[f], label=f"{f} (Spearman {r:.2f}, p={p:.3f})")
for a_, fila in TEMP.nlargest(5, f"{FUENTES[0]}_%").iterrows():
    ax[0].annotate(str(a_), (fila[INDICE_ENSO], fila[f"{FUENTES[0]}_%"]), fontsize=8, xytext=(4, 2), textcoords="offset points")
ax[0].axhline(100, color="k", lw=.5); ax[0].set_xlabel(f"{INDICE_ENSO} medio de la temporada (°C)"); ax[0].set_ylabel("% de la lluvia media de la temporada")
ax[0].legend(fontsize=8); ax[0].set_title(f"Temporada {TEMPORADA}")
orden = [c for c in TEMP.groupby("categoría")[INDICE_ENSO].mean().sort_values().index]
b = TEMP.groupby("categoría")[[f"{f}_%" for f in FUENTES]].median().reindex(orden)
b.plot.bar(ax=ax[1], color=[COL[f] for f in FUENTES], rot=30)
for k_, c_ in enumerate(orden): ax[1].text(k_, 5, f"n={int((TEMP['categoría'] == c_).sum())}", ha="center", fontsize=8)
ax[1].axhline(100, color="k", lw=.5); ax[1].set_ylabel("Mediana, % de la media"); ax[1].set_title("Lluvia de la temporada por categoría del evento")
fig.tight_layout(); fig.savefig(RES / f"RD1_01_temporada_enso_{S}.png", dpi=300); plt.show()
display(TEMP.sort_values(f"{FUENTES[0]}_%", ascending=False).head(10).round(1))
```

## 9. ¿Dónde llueve más en años de El Niño?

Mapa de la anomalía de la lluvia de la temporada en los años de El Niño (índice medio de la temporada en fase cálida), celda por celda de ERA5. Muestra qué parte de su zona es más sensible: insumo directo para el capítulo de exposición.

```python
inv = json.loads((DATOS / "inventario.json").read_text(encoding="utf-8"))
EM = xr.open_dataset(DATOS / inv["archivos"]["era5_mensual"])["pr"]
EM = EM.assign_coords(time=EM.indexes["time"].to_period("M").to_timestamp())
cruza = MESES_T[0] > MESES_T[-1]
anio_t = (EM.time.dt.year + ((EM.time.dt.month >= MESES_T[0]) & cruza).astype(int)).rename("temporada")
sel = EM.where(EM.time.dt.month.isin(MESES_T), drop=True)
tot = sel.groupby(anio_t.sel(time=sel.time)).sum("time", min_count=1)
umbral = enso.UMBRALES_ICEN["cálida débil"] if INDICE_ENSO == "ICEN" else enso.UMBRAL_ONI
ANIOS_NINO = [a for a, v in TEMP[INDICE_ENSO].items() if v >= umbral and a in tot["temporada"].values]
media = tot.sel(temporada=[a for a in TEMP.index if ANOM_INI <= a <= ANOM_FIN]).mean("temporada")
MAPA = 100 * (tot.sel(temporada=ANIOS_NINO).mean("temporada") / media - 1)
fig, ax = plt.subplots(figsize=(6, 6))
MAPA.plot(ax=ax, cmap="BrBG", center=0, cbar_kwargs={"label": "% sobre la media"})
gpd.GeoSeries([GEOM]).boundary.plot(ax=ax, color="k", lw=1)
ax.set_title(f"ERA5: lluvia {TEMPORADA} en años de El Niño ({INDICE_ENSO})" + chr(10) + ", ".join(map(str, ANIOS_NINO)), fontsize=9)
fig.savefig(RES / f"RD1_01_mapa_nino_{S}.png", dpi=300); plt.show()
```
